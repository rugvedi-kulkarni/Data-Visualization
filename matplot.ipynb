```python
# Real data

# Oscar winner movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

# plot
plt.plot(years, oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Year")
plt.ylabel("Revenues {in $M}")
```

```python
non_oscar_movies = ["Slumdog", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, 'o--r', label="Oscar Movies")
plt.plot(years, non_oscar_revenue,'<-y', label="Non Oscar Movies")

plt.title("Oscar Movies vs Non Oscar Movies Revenue")
plt.xlabel("years")
plt.ylabel("Revenues {in $M}")
plt.legend()

# outliers = extreme best or worst instead of normal results
# format  = [marker][string][color]

# if we want to control the indivisual formatting
# color="green"
# marker="o", "<"
# linestyle="--", "-."
# linewidth=5
```

```python
# properties and methods of plot

plt.plot(years, oscar_revenue,  label="Oscar Movies")
plt.plot(years, non_oscar_revenue,label="Non Oscar Movies")

plt.title("Oscar Movies vs Non Oscar Movies Revenue")
plt.xlabel("years")
plt.ylabel("Revenues {in $M}")
plt.legend()

print(plt.style.available) # to print all style of a plot
# plt.style.use('dark_background') # to set a style to the plot
plt.style.use("default")
plt.grid()
```

```python
# xkcd - cartonish style

with plt.xkcd():
    plt.plot(years, oscar_revenue,  label="Oscar Movies")
    plt.plot(years, non_oscar_revenue,label="Non Oscar Movies")
    plt.title("Oscar Movies vs Non Oscar Movies Revenue")
    plt.xlabel("years")
    plt.ylabel("Revenues {in $M}")
    plt.legend()
    plt.show()
```
